### 범주형 컬럼 인코딩

```python
# train 데이터를 복사 (원본 보존)
train_df2 = train_df.copy()
test_df2 = test_df.copy()
```

```python
# train 데이터에서 Group 컬럼 제거 (VIF 분석용)
train_df2.drop(columns=['Segment'], inplace=True)
```

```python
# 인코딩할 범주형 컬럼 목록
categorical_columns = [
    '할인건수_R3M', '방문횟수_앱_R6M', '방문횟수_PC_R6M', '방문일수_PC_R6M',
    '이용메뉴건수_ARS_R6M', '인입횟수_ARS_R6M', '할인건수_B0M', 'Life_Stage'
]

# 각 범주형 컬럼 factorize
for col in categorical_columns:
    # train 기준으로 인코딩
    train_df2[col], uniques = train_df2[col].factorize()

    # test에도 동일한 매핑 적용
    mapping = {category: code for code, category in enumerate(uniques)}

    # test 데이터 매핑
    test_df2[col] = test_df2[col].map(mapping)

    # test에 없는 새로운 값 처리 (예: NaN이면 -1로)
    test_df2[col] = test_df2[col].fillna(-1).astype(int)

# 결과 확인
display(train_df2.head())
display(test_df2.head())
```

# Train_VIF

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
import statsmodels.api as sm

# '기준년월'과 'ID' 컬럼 제외한 나머지 컬럼을 선택
vif_X = train_df2.drop(['기준년월', 'ID'], axis=1)

# 상수항 추가 (VIF 계산을 위한 준비)
vif_X2 = sm.add_constant(vif_X)

# VIF 계산
vif = pd.DataFrame()
vif['변수'] = vif_X2.columns
vif['VIF'] = [variance_inflation_factor(vif_X2.values, i) for i in range(vif_X2.shape[1])]

# 결과 출력
vif
```

```python
# pandas 출력 옵션 설정
pd.set_option('display.max_rows', None)  # 모든 행을 출력

# VIF 값을 기준으로 내림차순 정렬
vif_sorted = vif.sort_values(by='VIF', ascending=False)

# VIF 결과 출력
print(vif_sorted)
```
