```python
import pandas as pd
import glob
```

```python
# 경로 지정
path = '/content/drive/MyDrive/파이널 프로젝트/C_D_NotCD/data/'
file_list = glob.glob(path + 'selected(C_D_NotCD)*.csv')

print("파일 리스트 확인:", file_list)

if len(file_list) == 0:
    print("❗ 파일이 없습니다. 경로 또는 패턴을 확인하세요.")
else:
    merged_df = None
    for i, file in enumerate(file_list):
        df = pd.read_csv(file)
        print(f"Loaded: {file.split('/')[-1]} Shape: {df.shape}")

        if merged_df is None:
            merged_df = df
        else:
            merged_df = pd.merge(merged_df, df, on=['기준년월', 'ID'], how='inner')

    print("✅ 최종 병합된 데이터 shape:", merged_df.shape)
    display(merged_df.head())
```

```python
merged_df.columns
```

```python
# 컬럼 제거
selected_df = merged_df.drop(columns=['거주시도명'])

# 컬럼 개수와 shape 확인
print(selected_df.shape)
print(selected_df.columns)
```

```python
# 저장 경로와 파일명
file_path = '/content/drive/MyDrive/파이널 프로젝트/C_D_NotCD/data/Seleted(C_D_NotCD)_delecteE_all_train.parquet'

# parquet 파일로 저장
selected_df.to_parquet(file_path, index=False)

print("✅ parquet 파일로 저장 완료!")
```
