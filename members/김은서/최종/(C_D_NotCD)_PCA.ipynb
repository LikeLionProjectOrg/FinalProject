```python
# 기본
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 경고 뜨지 않게 설정
import warnings
warnings.filterwarnings('ignore')
```

```python
# 그래프 설정
sns.set()

plt.rcParams['figure.figsize'] = 12, 6
plt.rcParams['font.size'] = 14

# 데이터 전처리 알고리즘
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler

# 학습용과 검증용으로 나누는 함수
from sklearn.model_selection import train_test_split


# 차원 축소
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# 학습 모델 저장을 위한 라이브러리
import pickle
```

```python
# 데이터를 읽어온다.
df = pd.read_parquet('Seleted(C_D_NotCD)_delecteE_all_train.parquet')
df
```

```python
print(df.columns)
```

```python
# pca 사용
# n_components : 축소될 결과의 컬럼 개수
from sklearn.decomposition import PCA
pca = PCA(n_components=20)
# 결과 데이터를 제거한다.
X = df.drop(['기준년월', 'ID', 'Group'], axis=1)
# 학습한다.
pca.fit(X)
# 데이터를 변환한다.
r1 = pca.transform(X)
r1
```

```python
# 결과 데이터를 데이터프레임으로 변환
pca_df = pd.DataFrame(r1, columns=[f'p{i+1}' for i in range(r1.shape[1])])

# Segment 컬럼을 원본 df에서 가져와서 붙이기
pca_df['Group'] = df['Group'].values

# 결과 확인
pca_df
```

```python
pca_df['기준년월']=df['기준년월']
pca_df['ID']=df['ID']
pca_df
```

### train 저장

```python
pca_df.to_csv('(C_D_NotCD)_pca(20)_train_final.csv', index=False, encoding='utf-8-sig')
```

## Test

```python
df_test = pd.read_parquet('Seleted(C_D_NotCD)_all_test.parquet')
```

```python
df1 = pd.read_csv('data/test_model3_notE.csv')
```

```python
filtered_ids = df1[['ID', '기준년월']]
```

```python
# filtered_ids가 [(ID1, 기준년월1), (ID2, 기준년월2), ...] 형태라고 가정
filtered_df = pd.DataFrame(filtered_ids, columns=['ID', '기준년월'])

# df_test와 ID & 기준년월 기준으로 inner merge
final_df_test = df_test.merge(filtered_df, on=['ID', '기준년월'], how='inner')
```

```python
final_df_test
```

```python
# 불필요한 컬럼 제거
X_test = final_df_test.drop(['기준년월', 'ID'], axis=1)
X_test
```

```python
# train 데이터에서 feature 컬럼 리스트 확보
feature_columns = X.columns.tolist()

# test 데이터에서 동일한 컬럼, 동일한 순서로 선택
X_test = final_df_test[feature_columns]
```

```python
X_test
```

```python
# transform
X_test_pca = pca.transform(X_test)

# DataFrame 변환
pca_columns = [f'p{i+1}' for i in range(X_test_pca.shape[1])]
pca_df_test = pd.DataFrame(X_test_pca, columns=pca_columns)

# ID 붙이기
pca_df_test['ID'] = final_df_test['ID'].values
```

```python
pca_df_test
```
