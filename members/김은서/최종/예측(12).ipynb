```python
# 예측할 데이터를 읽어온다.
df1 = pd.read_csv('data/test_model3_notE.csv')
df1 = df1.drop(columns=['ID','기준년월'])
df1
```

```python
X_test = df1.drop('Segment', axis=1, errors='ignore')
X_test = X_test[scaler_columns]
```

```python
X_test
```

```python
# 예측할 데이터 변환
X_test['연령'] = Encoder1.transform(X_test['연령'])
```

```python
# 데이터 표준화
scaled_data = scalerX.transform(X_test)
scaled_data
```

```python
# 입력데이터를 test_X 변수에 담아준다.
X_text2 = scaled_data
```

```python
# 예측한다.
y_pred = model6.predict(X_text2)
y_pred_labels = le.inverse_transform(y_pred)
y_pred_labels
```

```python
# 결과를 붙힌다음 저장한다.
# 예측할 데이터를 다시 불러온다.
result_df = pd.read_csv('data/test_model3_notE.csv')
result_df['Segment'] = y_pred_labels
```

```python
result_df2 = result_df[['ID', 'Segment']]
```

```python
result_df2['Segment'].value_counts()
```

```python
most_common_by_id = result_df2.groupby('ID')['Segment'] \
    .agg(lambda x: x.mode().iloc[0]) \
    .reset_index()

# 컬럼명 확인
most_common_by_id.columns = ['ID', 'Segment']
```

```python
most_common_by_id
```

```python
most_common_by_id['Segment'].value_counts()
```

```python
most_common_by_id2 = most_common_by_id[most_common_by_id['Segment'] == 'not CD']
```
