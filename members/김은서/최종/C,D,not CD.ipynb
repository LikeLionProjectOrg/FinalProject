```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency, f_oneway

pd.set_option('display.float_format', '{:.5f}'.format)

# Cramér's V
def cramers_v(confusion_matrix):
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    r, k = confusion_matrix.shape
    return np.sqrt(chi2 / (n * (min(k, r) - 1)))

# Eta² from ANOVA
def eta_squared_anova(groups, overall_mean):
    ss_between = sum(len(g) * (g.mean() - overall_mean) ** 2 for g in groups)
    ss_total = sum(((x - overall_mean) ** 2 for g in groups for x in g))
    return ss_between / ss_total if ss_total != 0 else np.nan

# 효과 크기 계산 함수
def effect_sizes(df):
    num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    cat_cols = df.select_dtypes(exclude=[np.number]).columns.tolist()

    num_cols = [col for col in num_cols if col != 'Segment']
    cat_cols = [col for col in cat_cols if col != 'Segment']

    anova_results = []
    chi_results = []

    for col in df.columns:
        if col == 'Segment':
            continue

        # 수치형 변수
        if col in num_cols:
            data = df[[col, 'Segment']].dropna()
            groups = [group[col].values for name, group in data.groupby('Segment')]
            if len(groups) < 2:
                continue
            try:
                f_stat = f_oneway(*groups).statistic
                overall_mean = data[col].mean()
                eta2 = eta_squared_anova(groups, overall_mean)
                anova_results.append({
                    '변수': col, '유형': '수치형',
                    '계수종류': 'Eta²',
                    '상관계수': eta2
                })
            except:
                continue

        # 범주형 변수
        elif col in cat_cols:
            contingency = pd.crosstab(df[col], df['Segment'])
            if contingency.shape[0] > 1 and contingency.shape[1] > 1:
                try:
                    v = cramers_v(contingency)
                    chi_results.append({
                        '변수': col, '유형': '범주형',
                        '계수종류': "Cramér's V",
                        '상관계수': v
                    })
                except:
                    continue

    if anova_results:
        result_df1 = pd.DataFrame(anova_results).sort_values(by='상관계수', ascending=False).reset_index(drop=True)
    else:
        result_df1 = pd.DataFrame(columns=['변수', '유형', '계수종류', '상관계수'])

    if chi_results:
        result_df2 = pd.DataFrame(chi_results).sort_values(by='상관계수', ascending=False).reset_index(drop=True)
    else:
        result_df2 = pd.DataFrame(columns=['변수', '유형', '계수종류', '상관계수'])

    return result_df1, result_df2
```
