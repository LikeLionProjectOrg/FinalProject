```python
# 전체 클래스 순서 (A,B,C,D,E)
classes = ['A', 'B', 'C', 'D', 'E']  # 최종 클래스 순서

# ab 모델 숫자 → 문자 매핑
ab_map = {0:'A', 1:'B'}

# cde 모델 숫자 → 문자 매핑
cde_map = {0:'C', 1:'D', 2:'E'}

# 결과를 담을 배열 초기화
final_proba = np.zeros((df1.shape[0], len(classes)))
```

```python
threshold = 0.995  # A/B 확률 기준

final_pred = []  # 최종 결과 담을 리스트

for i in range(df1.shape[0]):
    # A/B 모델의 예측 확률 중 최대값과 해당 클래스
    ab_max_idx = np.argmax(proba_ab[i])
    ab_max_prob = proba_ab[i][ab_max_idx]
    ab_class = ab_map[ab_max_idx]  # 'A' 또는 'B'

    if ab_max_prob >= threshold:
        final_pred.append(ab_class)  # 신뢰 높으면 A/B 사용
    else:
        # 그렇지 않으면 C/D/E 모델의 결과 사용
        cde_max_idx = np.argmax(proba_cde[i])
        cde_class = cde_map[cde_max_idx]
        final_pred.append(cde_class)
```

```python
ab_weight = 0.255  # 0.1보다는 조금 크게
cde_weight = 0.745

final_proba = np.zeros((df1.shape[0], len(classes)))

for i, cls_num in enumerate(model_ab.classes_):
    cls_str = ab_map[cls_num]
    idx = classes.index(cls_str)
    final_proba[:, idx] = proba_ab[:, i] * ab_weight

for i, cls_num in enumerate(model_cde.classes_):
    cls_str = cde_map[cls_num]
    idx = classes.index(cls_str)
    final_proba[:, idx] = proba_cde[:, i] * cde_weight

predicted_idx = np.argmax(final_proba, axis=1)
final_pred = [classes[i] for i in predicted_idx]
```
