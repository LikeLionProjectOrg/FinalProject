1. 카드 이용 관련 비율/평균 
- 총 금액/건수 = 1건당 금액

```python
train_df['신용_1건당이용금액'] = train_df['이용금액_R3M_신용'] / (train_df['이용건수_신용_R3M'] + 1e-5)
```

- 최근 3개월 vs 12개월 = 소비집중도

```python
train_df['소비집중도_신용'] = train_df['이용금액_일시불_R3M'] / (train_df['이용금액_일시불_R12M'] + 1e-5)
```

- 온라인 vs 오프라인 비율 

```python
train_df['온라인비중'] = train_df['이용금액_온라인_B0M'] / (
    train_df['이용금액_온라인_B0M'] + train_df['이용금액_오프라인_B0M'] + 1e-5)
```

2. 업종별, 쇼핑별 총합 

```python
train_df['쇼핑업종총합'] = train_df['_1순위쇼핑업종_이용금액'] + train_df['_2순위쇼핑업종_이용금액'] + train_df['_3순위쇼핑업종_이용금액']
train_df['업종이용금액총합'] = train_df['_1순위업종_이용금액'] + train_df['_2순위업종_이용금액'] + train_df['_3순위업종_이용금액']
```

3. 최근 시점 변화량/
0M vs 2M (이번 달 vs 두달 전)


```python
train_df['정상청구원금_증감'] = train_df['정상청구원금_B0M'] - train_df['정상청구원금_B2M']
train_df['일시불잔액_증감'] = train_df['잔액_일시불_B0M'] - train_df['잔액_일시불_B2M']
```

4. 이용패턴 관련 조합 
- 총 이용카드수 대비 이용가맹점수 -> 다양성지표

```python
train_df['가맹점_카드당비율'] = train_df['이용가맹점수'] / (train_df['이용카드수_신용체크'] + 1e-5)
```

- 상향가능한도금액 vs 이용금액

```python
train_df['한도대비이용'] = train_df['이용금액_R3M_신용체크'] / (train_df['상향가능한도금액'] + 1e-5)
```

5. 이진 변수 묶기(수신거부여부 등)

```python
train_df['수신거부합계'] = train_df[['수신거부여부_TM', '수신거부여부_메일', '수신거부여부_DM']].sum(axis=1)
```

6. 특이 행동 탐지
- 페이_온라인_B0M 유무 -> 간편결제 여부

```python
train_df['페이사용여부'] = (train_df['이용금액_페이_온라인_B0M'] > 0).astype(int)
```

```python
def make_features(df):
    # 1건당 결제 금액
    df['신용_1건당이용금액'] = df['이용금액_R3M_신용'] / (df['이용건수_신용_R3M'] + 1e-5)

    # 소비 집중도 (최근 3개월 vs 12개월)
    df['소비집중도_신용'] = df['이용금액_일시불_R3M'] / (df['이용금액_일시불_R12M'] + 1e-5)

    # 온라인 비중
    df['온라인비중'] = df['이용금액_온라인_B0M'] / (
        df['이용금액_온라인_B0M'] + df['이용금액_오프라인_B0M'] + 1e-5)

    # 업종별 합계
    df['쇼핑업종총합'] = df['_1순위쇼핑업종_이용금액'] + df['_2순위쇼핑업종_이용금액'] + df['_3순위쇼핑업종_이용금액']
    df['업종이용금액총합'] = df['_1순위업종_이용금액'] + df['_2순위업종_이용금액'] + df['_3순위업종_이용금액']

    # 변화량 계산
    df['정상청구원금_증감'] = df['정상청구원금_B0M'] - df['정상청구원금_B2M']
    df['일시불잔액_증감'] = df['잔액_일시불_B0M'] - df['잔액_일시불_B2M']

    # 카드당 가맹점 수
    df['가맹점_카드당비율'] = df['이용가맹점수'] / (df['이용카드수_신용체크'] + 1e-5)

    # 한도 대비 이용
    df['한도대비이용'] = df['이용금액_R3M_신용체크'] / (df['상향가능한도금액'] + 1e-5)

    # 수신거부 합계
    df['수신거부합계'] = df[['수신거부여부_TM', '수신거부여부_메일', '수신거부여부_DM']].sum(axis=1)

    # 페이 사용 여부
    df['페이사용여부'] = (df['이용금액_페이_온라인_B0M'] > 0).astype(int)

    return df
```

```python
train_df = make_features(train_df)
test_df = make_features(test_df)
```
